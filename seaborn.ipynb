# 🌊 Seaborn

## **Seaborn is Built on Top of Matplotlib**

- High-level interface with beautiful default styles.
- Simplifies complex plots (like **Box Plot**, **Heatmap**, etc.).
- Integrates well with **Pandas DataFrames**.
- Less code, more aesthetics.

---

## 🧠 Why Use Seaborn?

Matplotlib is powerful, but sometimes it requires more code to create attractive visualizations.

Seaborn provides:

- Better-looking graphs by default.
- Easier syntax for statistical visualizations.
- Better integration with Pandas.
- Built-in themes and color palettes.

---

## 🎯 Memory Rule

> **Beautiful Statistical Graphs → Seaborn**

If you want attractive statistical plots with less code, think:

```text
Need Beautiful Statistical Visualization?

↓

Seaborn
```

 # 🔗 Relational Plots

## **Definition**

Relational plots show relationships between **two or more numerical variables**.

---

## **Usage**

- Used to visualize relationships in datasets (e.g., correlation between **age** and **medical expenses**).
- Helps understand **trends, patterns, and clusters** in numerical data.

---

## **Characteristics**

- Typically uses **Scatter Plots (`sns.scatterplot`)** and **Line Plots (`sns.lineplot`)**.
- Supports **hue**, **size**, and **style** to add more insights.

---

## **Applications**

### 📡 Telecommunication
Relationship between **call duration** and **customer satisfaction**.

### 💰 Finance
Relationship between **income** and **expenditure**.

### 🛍️ Retail
Impact of **discount percentage** on **sales**.

---

## 🧠 When Should You Think of a Relational Plot?

```text
Question asks:

Relationship Between Variables

↓

Trend

↓

Pattern

↓

Correlation

↓

Relational Plot
```

---

## 🎯 Memory Rule

> **Relationship Between Two or More Numerical Variables → Relational Plot**

---

## 📌 Common Seaborn Relational Plots

| Plot | Function | Best Used For |
|------|----------|---------------|
| Scatter Plot | `sns.scatterplot()` | Relationship between two numerical variables |
| Line Plot | `sns.lineplot()` | Trends over time or ordered data |

---

## 💡 Extra Features

| Parameter | Purpose |
|-----------|---------|
| `hue` | Differentiate categories using colors |
| `size` | Represent another variable using marker size |
| `style` | Represent categories using different marker styles |

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
df = pd.read_csv("telecom_customer_data.csv")
df.head()
````

````python
df.shape
````

````python
df.info()
````

````python
# Create a scatter plot
plt.figure(figsize = (8,5))
sns.scatterplot(data = df , x = "Call_Duration", y = "Satisfaction_Score", hue ="Network_Provider")

#Now the legend outside the plot
plt.legend(title = "Network Provider", bbox_to_anchor = (1,1), loc = "upper left")

#Add Labels and title
plt.title("Call Duration vs Satisfaction Score")
plt.xlabel("Call Duration (minutes)")
plt.ylabel("Satisfaction Score")

#Show the plot
plt.show()
````

# 📊 Categorical Plots

## **Definition**

Categorical plots display relationships involving **categorical variables** (e.g., **gender, department, product category**).

---

## **Usage**

- Used to analyze **grouped data** (e.g., average number of calls per customer segment).
- Helps in comparing different categories.

---

## **Characteristics**

- Includes **Bar Plots (`sns.barplot`)**, **Box Plots (`sns.boxplot`)**, **Violin Plots (`sns.violinplot`)**, etc.
- Allows **comparison across multiple groups**.

---

## **Applications**

### 📡 Telecommunication
Distribution of **customer complaints** by **network provider**.

### 🏥 Healthcare
Distribution of **patient recovery time** by **disease**.

### 🎓 Education
Comparison of **students' performance** by **subject**.

# Categorical Plot : Count of Customers by Network Provider

````python
plt.figure(figsize = (6 , 4))
sns.countplot(data = df , x = "Network_Provider" , hue = "Network_Provider" , palette = "viridis")
plt.title("Customer Distribution by Network Provider")
plt.xlabel("Network Provider")
plt.ylabel("Count of customers")
plt.legend().remove()
plt.show()
````

# 📈 Distribution Plots

## **Definition**

Distribution plots visualize the **spread** and **shape** of data.

---

## **Usage**

- Helps understand **data distribution**, **skewness**, and **outliers**.
- Useful for **identifying trends** and **anomalies**.

---

## **Characteristics**

- Includes **Histograms (`sns.histplot`)**, **KDE Plots (`sns.kdeplot`)**, and **Rug Plots (`sns.rugplot`)**.
- Displays **density estimates** and **probability distributions**.

---

## **Applications**

### 📡 Telecommunication
Distribution of **call durations** per customer.

### 💰 Finance
Distribution of **transaction amounts** in a bank.

### 🛒 E-commerce
Analyzing **purchase frequency** of products.

# Distribution Plot: Distribution of call Durations

````python
plt.figure(figsize=(7,4))
sns.histplot(df["Call_Duration"], bins = 20,  kde = True, color = "blue", edgecolor = "black")
plt.title("Distribution of Call Durations")
plt.xlabel("Call Duration (minutes)")
plt.ylabel("Frequency")
plt.show()                 #Outlier -> Far away from mean 
````

# 🔥 Matrix Plots

## **Definition**

Matrix plots visualize **relationships between multiple variables** using **heatmaps** and **clustering methods**.

---

## **Usage**

- Helps find **correlations** and **patterns** in datasets.
- Used in **feature selection** and **anomaly detection**.

---

## **Characteristics**

- Includes **Heatmaps (`sns.heatmap`)** and **Cluster Maps (`sns.clustermap`)**.
- Displays **intensity** and **similarity** between variables.

---

## **Applications**

### 📡 Telecommunication
Heatmap of **network congestion** across different locations.

### 💰 Finance
Correlation between **various stock indices**.

### 🌦️ Weather Analytics
Heatmap of **temperature variations** over time.

````python
plt.figure(figsize = (8 , 6))
numeric_df = df.select_dtypes(include = ['number']) #select only numeric columns
sns.heatmap(numeric_df.corr() , annot = True, cmap = "coolwarm"  , fmt = ".2f", linewidths = 0.5)
plt.title("Correlation Matrix of features")
plt.show()
````

# Box Plot: Satisfaction Score by Network Provider

````python
plt.figure(figsize= (8 , 5))
sns.boxplot(data = df , x = "Network_Provider" , y = "Satisfaction_Score")
plt.title("Satisfaction Score Across Different Service Providers")
plt.xlabel("Network Provider")
plt.ylabel("Satisfaction Score")
plt.show()
````

# Violin Plot: Call Duration by Network Provider 

````python
plt.figure(figsize= (8 , 5))
sns.violinplot(
    data = df,
    x = ("Network_Provider"),
    y = ("Satisfaction_Score")
)
plt.title("Call Duration by data Network providers")
plt.xlabel("Network Provider")
plt.ylabel("Call Duration (minutes)")
plt.show()
````

# Used Car Price using Matplotlib and Seaborn
Dataset: Used Car Listings From CarDekho
Objective: Solve business problems visually - price trends, feature effects. resale value patterns.

Step 1: Import & Configure 

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
# Set Theme
sns.set(style = "whitegrid")
plt.rcParams ['figure.figsize'] = (12 , 8)
````

````python
# Load the Dataset

df = pd.read_csv("CAR_DETAILS_FROM_CAR_DEKHO.csv")
df.head()
````

````python
df.shape
````

````python
df.info()
````

````python
df.describe()
````

# Step 3: Fuel Type Distribution

````python
# Fuel Type Availability
sns.countplot(data = df , x = "fuel" , hue = "fuel", palette = "Set2" , legend = False)
plt.title("Fuel Type Distribution")
plt.show()
````

# inference / Observation
### 🔍 Observation:

- Petrol cars dominate the listings with approximately **2100 entries**.
- Diesel cars follow, around **2200 entries**.
- Very few listings for **CNG/LPG (~100)**.
- **No electric vehicles** are listed in this dataset.

### 📌 Inference:

This indicates that the **resale market is heavily driven by petrol and diesel vehicles**.

The **absence of electric vehicles** shows either **very low adoption** or that they are **too new for resale**.

This also reflects the **current composition of India's used car market**.

# Step 4: Selling Price Distribution

````python
# Price Range Analysis
sns.histplot(df["selling_price"] , kde = True , bins = 30 , color = "skyblue")
plt.title("Selling Price Distribution")
plt.xlabel("Price (INR)")
plt.show()
````

# Inference / observation
### 🔍 Observation:

- Most cars fall under the **₹5 lakh** price range.
- A **long tail** exists with premium cars priced **₹10 lakh and above**.

### 📌 Inference:

The used car market is **skewed toward budget buyers**. A few luxury vehicles raise the maximum price, but the bulk of demand lies in the **affordable segment**.

# Step 5: Price by Fuel Type (Boxplot)

````python
sns.boxplot(data = df , x = "fuel" , y = "selling_price" , palette = "Set1" )
plt.title("Price Distribution by Fuel Type")
plt.show()
````

# Inference / Observation
### 🔍 Observation:

- Diesel vehicles have **higher median selling prices**.
- Petrol cars have a **wider spread** but **lower median**.

### 📌 Inference:

Diesel cars often hold **better resale value** due to **fuel efficiency** and **higher initial price**. CNG vehicles show **lower prices**, often used for **commercial or city travel**.

# Step 6: Price by Transmission

````python
sns.boxplot(data = df , x = "transmission" , hue = "transmission" , y = "selling_price" , palette = "coolwarm" )
plt.title("Selling Type by Transmission Type")
plt.show()
````

# Inference / Observation
### 🔍 Observation:

- Automatic cars show a **slightly higher median price**.
- Manual cars are **more common** with **greater price variation**.

### 📌 Inference:

Automatic transmission vehicles have **higher resale value**, possibly due to **urban driving preferences** and **convenience**.

# Step 7: Price by Ownership History

````python
sns.violinplot(data = df , x = "owner" , hue = "owner" , y = "selling_price" , palette = "Set3")
plt.title("Price by Ownership History")
plt.show()
````

# Inference / Observation
### 🔍 Observation:

- **First-owner** cars have **higher resale prices**.
- Prices **drop significantly** with each additional owner.

### 📌 Inference:

Ownership history impacts **resale value**. **First-owner vehicles command a premium**, while **2nd and 3rd-owner** vehicles lose value rapidly

# Step 8: Correlation Heatmap

````python
correlation = df[["year" , "km_driven" , "selling_price"]].corr()
sns.heatmap(correlation, annot = True , cmap = "YlGnBu")
plt.title("Feature Correlation Heatmap")
plt.show()
````

# Inference / Observation
### 🔍 Observation:

- Selling price has a **mild positive correlation** with **`year`**.
- **Mild negative correlation** with **`km_driven`**.

### 📌 Inference:

While there are **no strong correlations**, the **age and usage** of a car still show the expected directional impact on resale value. **No major multicollinearity** issues for basic regression model
